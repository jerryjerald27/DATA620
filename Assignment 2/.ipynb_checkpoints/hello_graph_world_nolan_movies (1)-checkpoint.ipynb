# DATA 620 – Assignment 1: "Hello, Graph World"

## Overview
This notebook is a simple "beachhead" exercise to confirm the Python environment works. It builds a small network of **Christopher Nolan films and their recurring cast** using NetworkX and plots it with matplotlib.

## Data
`nolan_films_cast.csv` is a hand-compiled edge list with one row per film–actor pair.

| Column | Description |
|--------|-------------|
| `film` | Nolan film, from *Batman Begins* (2005) through *Oppenheimer* (2023) |
| `actor` | Actor who appears in that film (main or cameo role) |

Only actors who appear in at least two of the nine films are included, so this is not the full cast of each film.

## Graph structure
- **Nodes:** films and actors
- **Edges:** an actor is connected to each film they appear in

## 1. Setup

```python
import sys
import pandas as pd
import networkx as nx
import matplotlib
import matplotlib.pyplot as plt

print("Python    :", sys.version.split()[0])
print("pandas    :", pd.__version__)
print("NetworkX  :", nx.__version__)
print("matplotlib:", matplotlib.__version__)
```

## 2. Load the data

```python
df = pd.read_csv("nolan_films_cast.csv")

print("Rows:", len(df))
df.head()
```

## 3. Build the graph

Each row of the CSV becomes one edge between a film and an actor.

```python
G = nx.from_pandas_edgelist(df, source="film", target="actor")

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
```

## 4. Plot the graph

Films are red and actors are light blue.

```python
films = df["film"].unique()
colors = ["red" if node in films else "lightblue" for node in G]
pos = nx.spring_layout(G, k=0.6, seed=42)

plt.figure(figsize=(12, 8))
nx.draw(G, pos, node_color=colors, with_labels=True, node_size=900,
        font_size=8, edge_color="gray")
plt.title("Christopher Nolan Films and Their Recurring Cast")
plt.show()
```

## 5. Results

Number of films each actor appears in:

```python
df["actor"].value_counts().head(5)
```

**Takeaway:** Michael Caine appears in 7 of the 9 films and Cillian Murphy in 6, which makes them the most connected actors in the graph. The Batman trilogy films are tightly linked because they share most of the same cast.
